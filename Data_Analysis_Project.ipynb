**Remove Outliers**

We apply the IQR method on the `Sales` column and keep two versions of the data:

- `df`: filtered by the lower bound only. The lower bound is negative (-271.71) and no sale falls below it, so `df` is effectively the **original data** (all 9,994 rows).
- `df_new`: filtered by both bounds (Sales between -271.71 and 498.93), i.e. the **outlier-free data**.

This lets us compare the results before and after removing outliers.

```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(lower_bound)
print(upper_bound)

df_original = df.copy()

df_new = df[(df["Sales"] >= lower_bound) & (df["Sales"] <= upper_bound)].copy()
df = df[(df["Sales"] >= lower_bound)].copy()
df_new.head(5)
```

**3. Feature Engineering**

    • Create at least 2–3 new features

    • Explain why these features are useful

• Create at least 2–3 new features

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df_new['Order Date'] = pd.to_datetime(df_new['Order Date'])
df_new['Ship Date'] = pd.to_datetime(df_new['Ship Date'])

# Feature 1: Shipping Duration (in days)
df['Shipping_Duration'] = (df['Ship Date'] - df['Order Date']).dt.days
df_new['Shipping_Duration'] = (df_new['Ship Date'] - df_new['Order Date']).dt.days

# Feature 2: Profit Ratio
df['Profit_Ratio'] = df['Profit'] / df['Sales']
df_new['Profit_Ratio'] = df_new['Profit'] / df_new['Sales']

# Feature 3: Most Ordered Product
product_rank = df.groupby('Product Name')['Quantity'].sum().rank(ascending=False, method='dense')
df['product_rank'] = df['Product Name'].map(product_rank)

product_rank_new = df_new.groupby('Product Name')['Quantity'].sum().rank(ascending=False, method='dense')
df_new['product_rank'] = df_new['Product Name'].map(product_rank_new)

print(df[['Order Date', 'Ship Date', 'Shipping_Duration', 'Sales', 'Profit', 'Profit_Ratio' , 'product_rank']].head())
print("For new df \n", df_new[['Order Date', 'Ship Date', 'Shipping_Duration', 'Sales', 'Profit', 'Profit_Ratio' , 'product_rank']].head())

```

### Why are these features useful?

*   **Shipping Duration**: This feature helps in understanding the efficiency of the shipping process. Longer shipping durations might indicate bottlenecks in logistics or customer dissatisfaction. It can also be used to identify orders that are taking unusually long to ship.
*   **Profit Ratio**: This feature normalizes profit by sales, providing a clearer picture of the profitability of individual transactions or product categories. A low or negative profit ratio indicates that certain sales might not be financially viable.
*   **Most Ordered Product**: Identifying the most ordered product helps in understanding customer preferences and demand. This insight is crucial for inventory management, marketing strategies, and product development. It can highlight popular items that should be well-stocked or promoted.

**4. Data Analysis (GroupBy)**

    • Use groupby operations

    • Answer analytical questions such as:

        o Which category has the highest value?

        o What are the average values per group?

        o Any noticeable patterns?

• Use groupby operations (on `df`: the original data, since no sale was below the lower bound)

```python
# 1.Total Sales by Category
sales_by_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(f"Total Sales by Category:\n{sales_by_category}")

# 2.Average Sales, Profit, Discount per Category
avg_by_category = df.groupby("Category")[["Sales", "Profit", "Discount"]].mean().round(2)
print(f"\nAverage Values per Category:\n{avg_by_category}")

# 3.Total Profit by Region
profit_by_region = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)
print(f"\nTotal Profit by Region:{profit_by_region}")

# 4.Average Profit per Segment
avg_profit_segment = df.groupby("Segment")["Profit"].mean().sort_values(ascending=False)
print(f"\nAverage Profit per Segment:\n{avg_profit_segment}")

# 5.Top 10 Sub-Categories by Sales
top_sub = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(10)
print(f"\nTop 10 Sub-Categories by Sales:\n{top_sub}")

# 6.Avg Shipping Duration per Ship Mode
ship_avg = df.groupby("Ship Mode")["Shipping_Duration"].mean().round(2).sort_values()
print(f"\nAverage Shipping Duration per Ship Mode (days):\n{ship_avg}")

# 7.Average Profit Ratio per Category
profit_ratio_by_category = df.groupby("Category")["Profit_Ratio"].mean().round(3).sort_values(ascending=False)
print(f"\nAverage Profit Ratio per Category:\n{profit_ratio_by_category}")
```

• Answer analytical questions such as:

    o Which category has the highest value?

    o What are the average values per group?

    o Any noticeable patterns?

```python
# Q1:Which Category has the Highest Sales?
top_cat = sales_by_category.idxmax()
top_val = sales_by_category.max()
print(f"Category with Highest Total Sales\n{top_cat}  :  ${top_val}")
print("Reason: Technology products (phones, printers, machines) have higher unit prices compared to furniture or office supplies, which drives up their total sales revenue")
```

```python
# Q2:Average Values per Group (Category)
print(f"Average Sales & Profit per Category:\n{avg_by_category[['Sales', 'Profit', 'Discount']]}")
print("\nKey observations:")
print("Technology -> Highest average sales and highest profit per order")
print("Furniture -> Decent average sales but very weak profit due to high discounts")
print("Office Supplies -> Lowest average sales, but better profit than Furniture because discounts are slightly lower")
```

```python
# Q3:Noticeable Patterns
print("\nPattern 1: Region Performance")
print(profit_by_region)
print("West & East regions generate the most profit")

print("\nPattern 2: Technology = High Sales & High Profit")
print(avg_by_category)
print("Technology has the highest avg Sales and avg Profit per order")

print("\nPattern 3: Segment Profit is similar across groups")
print(avg_profit_segment)
print("Home Office slightly edges out Corporate and Consumer segments")
```
